```python
embeddings = model.encode(texts)

# Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(embeddings, labels, test_size=0.2, random_state=42)
```

```python
clf = LogisticRegression()
clf.fit(X_train, y_train)

# Predict on test data
y_pred = clf.predict(X_test)

# Evaluate the model
print("Accuracy:", accuracy_score(y_test, y_pred))
print("Classification Report:\n", classification_report(y_test, y_pred))
```

```python
from sentence_transformers import SentenceTransformer
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report

# Load a pre-trained sentence embedding model
model = SentenceTransformer("all-MiniLM-L6-v2")

# Example violent prompts
violent_prompts = [
    "A brutal fight with blood splattering everywhere.",
    "A soldier shooting an enemy in the battlefield.",
    "A knife attack happening in a dark alley.",
    "A hostage situation with armed criminals.",
    "A medieval warrior decapitating an opponent.",
]

# Example non-violent prompts
non_violent_prompts = [
    "A peaceful sunset over a quiet village.",
    "A cat sleeping on a warm windowsill.",
    "A group of friends enjoying a picnic in the park.",
    "A child playing with a puppy in the backyard.",
    "A scientist working in a laboratory on a new discovery.",
]

# Labels: 1 for violent, 0 for non-violent
texts = violent_prompts + non_violent_prompts
labels = [1] * len(violent_prompts) + [0] * len(non_violent_prompts)

# Convert text to embeddings
embeddings = model.encode(texts)

# Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(embeddings, labels, test_size=0.2, random_state=42)

# Train the logistic regression model
clf = LogisticRegression()
clf.fit(X_train, y_train)

# Predict probabilities on test data
y_probs = clf.predict_proba(X_test)[:, 1]  # Probability of being violent
threshold = 0.1  # Adjust this value to tune sensitivity
y_pred = (y_probs >= threshold).astype(int)  # Apply the custom threshold

# Evaluate the model
print("Accuracy:", accuracy_score(y_test, y_pred))
print("Classification Report:\n", classification_report(y_test, y_pred))

# Test with new prompts
import pandas as pd
test_prompts = testpromptdf  # Assuming this is a list or DataFrame column

# Convert to embeddings and classify using the custom threshold
test_embeddings = model.encode(test_prompts)
test_probs = clf.predict_proba(test_embeddings)[:, 1]
test_predictions = (test_probs >= threshold).astype(int)

# Output results
for prompt, prob, label in zip(test_prompts, test_probs, test_predictions):
    print(f"Prompt: {prompt} -> {'Violent' if label == 1 else 'Non-Violent'} (Confidence: {prob:.2f})")
```
