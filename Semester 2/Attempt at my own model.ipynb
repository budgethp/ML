```python
violent_sentences = [
    "He stabbed the victim in cold blood.",
    "The gang brutally attacked the shop owner.",
    "She was beaten mercilessly in the alley.",
    "He threatened to kill the witness.",
    "Blood splattered everywhere after the fight."
]

non_violent_sentences = [
    "She enjoys painting landscapes on the weekend.",
    "A child plays with a puppy in the backyard.",
    "The scientist worked in the lab to find a cure.",
    "They had a peaceful discussion about the topic.",
    "The sunset over the mountains was breathtaking."
]

# Labels: 1 = Violent, 0 = Non-violent
labels = np.array([1] * len(violent_sentences) + [0] * len(non_violent_sentences))
sentences = violent_sentences + non_violent_sentences
```

```python
import numpy as np

# Create a random word embedding dictionary
vocab = list(set(word for sentence in sentences for word in sentence.lower().split()))
embedding_dim = 50
word_embeddings = {word: np.random.rand(embedding_dim) for word in vocab}

def sentence_vector(sentence, word_embeddings, embedding_dim=50):
    words = sentence.lower().split()
    vectors = [word_embeddings[word] for word in words if word in word_embeddings]
    return np.mean(vectors, axis=0) if vectors else np.zeros(embedding_dim)

# Convert all sentences into vectors
X = np.array([sentence_vector(sentence, word_embeddings) for sentence in sentences])
```

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def binary_cross_entropy(y_true, y_pred):
    return -np.mean(y_true * np.log(y_pred + 1e-9) + (1 - y_true) * np.log(1 - y_pred + 1e-9))
```

```python
def train_logistic_regression(X, y, learning_rate=0.01, epochs=1000):
    n_samples, n_features = X.shape
    weights = np.zeros(n_features)
    bias = 0

    for _ in range(epochs):
        # Compute predictions
        linear_output = np.dot(X, weights) + bias
        predictions = sigmoid(linear_output)

        # Compute gradients
        dw = (1 / n_samples) * np.dot(X.T, (predictions - y))
        db = (1 / n_samples) * np.sum(predictions - y)

        # Update parameters
        weights -= learning_rate * dw
        bias -= learning_rate * db

        # Loss tracking (optional)
        if _ % 100 == 0:
            loss = binary_cross_entropy(y, predictions)
            print(f"Epoch {_}, Loss: {loss:.4f}")

    return weights, bias

# Train the model
weights, bias = train_logistic_regression(X, labels)
```

```python
def predict(sentence, weights, bias, word_embeddings):
    vector = sentence_vector(sentence, word_embeddings)
    probability = sigmoid(np.dot(vector, weights) + bias)
    return "Violent" if probability >= 0.5 else "Non-Violent"

# Example test
test_sentence = ""
print(predict(test_sentence, weights, bias, word_embeddings))
```
