```python
import numpy as np
import re
from collections import defaultdict

def preprocess(text):
    text = text.lower()  
    text = re.sub(r'[^a-z\s]', '', text)  
    return text.split()

def build_vocab(corpus):
    vocab = {}
    for sentence in corpus:
        for word in preprocess(sentence):
            if word not in vocab:
                vocab[word] = len(vocab)  
    return vocab

def build_cooccurrence_matrix(corpus, vocab, window_size=2):
    vocab_size = len(vocab)
    matrix = np.zeros((vocab_size, vocab_size))

    for sentence in corpus:
        words = preprocess(sentence)
        for i, word in enumerate(words):
            if word in vocab:
                word_idx = vocab[word]
                for j in range(max(0, i - window_size), min(len(words), i + window_size + 1)):
                    if i != j and words[j] in vocab:
                        context_idx = vocab[words[j]]
                        matrix[word_idx, context_idx] += 1
    return matrix

def reduce_dimensions(matrix, dim=2):
    U, S, Vt = np.linalg.svd(matrix)
    return U[:, :dim]
```

```python
def cosine_similarity(vec1, vec2):
    return np.dot(vec1, vec2) / (np.linalg.norm(vec1) * np.linalg.norm(vec2))

def detect_nsfw(prompt, vocab, word_vectors, nsfw_keywords):
    prompt_words = preprocess(prompt)
    
    # Check if any NSFW words exist in the prompt directly
    for word in prompt_words:
        if word in nsfw_keywords:
            return True  # Direct match found

    # Compare prompt word vectors to NSFW word vectors
    nsfw_indices = [vocab[word] for word in nsfw_keywords if word in vocab]
    prompt_indices = [vocab[word] for word in prompt_words if word in vocab]

    for p_idx in prompt_indices:
        for nsfw_idx in nsfw_indices:
            if cosine_similarity(word_vectors[p_idx], word_vectors[nsfw_idx]) > 0.8:
                return True  # Similar meaning detected

    return False  # Safe prompt
```

```python
corpus = [
    "I love drawing beautiful landscapes",
    "This artwork depicts a battle scene",
    "An explicit nude painting",
    "A fantasy warrior with a sword",
    "A brutal murder scene with blood",
    "Romantic couple in a sunset"
]

# Step 1: Build Vocabulary & Word Vectors
vocab = build_vocab(corpus + nsfw_keywords)
co_matrix = build_cooccurrence_matrix(corpus, vocab, window_size=2)
word_vectors = reduce_dimensions(co_matrix, dim=2)

# Test Cases
prompts = [
    "Draw a beautiful naked woman",
    "Show a brutal battle with blood everywhere",
    "A peaceful scenery with flowers",
    "Illustrate a cute puppy playing in the park"
]

for prompt in prompts:
    is_nsfw = detect_nsfw(prompt, vocab, word_vectors, nsfw_keywords)
    print(f"Prompt: '{prompt}' -> NSFW: {is_nsfw}")
```
