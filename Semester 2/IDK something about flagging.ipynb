```python
from sentence_transformers import SentenceTransformer, util

model = SentenceTransformer('all-MiniLM-L6-v2')

violent_examples = ["A person being stabbed", "Gunfight with blood everywhere", "A person violently dying", "A person being dismembered"]
violent_embeddings = model.encode(violent_examples)

def is_violent(prompt):
    prompt_embedding = model.encode(prompt)
    scores = util.cos_sim(prompt_embedding, violent_embeddings)
    return scores.max().item() > 1  # Threshold

prompt = "A brutal battle scene with blood and swords"
print(is_violent(prompt))  # True or False
```
